## 🧪 Step 1 — Basic Flask App


````python
from flask import Flask, request, jsonify

app = Flask(__name__)

@app.route('/')
def home():
    return {"status": "GenAI API is running!"}
````

## 🧪 Step 2 — Add LLM Integration (Mock)
This uses a placeholder LLM function.


````python
def mock_llm(prompt):
    return f"Processed: {prompt[:50]}..."
````

## 🧪 Step 3 — Create `/generate` Route


````python
@app.route('/generate', methods=['POST'])
def generate():
    data = request.get_json()
    text = data.get('text')
    if not text:
        return jsonify({"error": "Missing 'text' field."}), 400
    output = mock_llm(text)
    return jsonify({"result": output})
````

## 🧪 Exercise 6 — Add Logging
Modify your Flask app so that each request logs:
- timestamp
- input text
- output


````python
import datetime

def log_request(input_text, output):
    timestamp = datetime.datetime.now().isoformat()
    print(f"[{timestamp}] INPUT: {input_text}")
    print(f"[{timestamp}] OUTPUT: {output}")
````

# 🔥 Bonus Section — Running Free Local LLMs with Ollama
This section shows you how to use open-source LLMs *locally* with no API keys.

You can run models like:
- Llama 3
- Mistral
- Phi-3
- DeepSeek-R1

All **offline** and **free**.


## 🟦 Step 1 — Install Ollama
### macOS / Linux
```bash
curl -fsSL https://ollama.com/install.sh | sh
```

### Windows
Download from: https://ollama.com


## 🟩 Step 2 — Run Your First Free Model
Example with **Llama 3**:
```bash
ollama run llama3
```

Example with **Mistral**:
```bash
ollama run mistral
```

Example with **Phi-3** (tiny & fast):
```bash
ollama run phi3
```


## 🟨 Step 3 — Use Local LLMs in Python
Make sure Ollama is running, then use:


````python
from flask import requests

prompt = "Explain RAG in 2 sentences."

response = requests.post(
    "http://localhost:11434/api/generate",
    json={"model": "llama3", "prompt": prompt}
)

print(response.json()["response"])
````
