## 🔹 What Is Prompt Engineering?
Prompt engineering is the process of designing effective inputs for AI models.

A good prompt includes:
- **Role**
- **Context**
- **Task / Instructions**
- **Constraints**
- **Examples** (optional)


## 🔹 Example: Weak vs Improved Prompt
### ❌ Weak
"Explain artificial intelligence."

### ✅ Improved
```
You are a professor of computer science.
Context: The student is new to AI.
Task: Explain artificial intelligence.
Constraints: Use simple language. Keep it under 80 words.
```
