## 🔹 What Is RAG?
RAG = Retrieval-Augmented Generation.

It improves AI accuracy using this pipeline:
1. Chunk text
2. Convert chunks to embeddings
3. Store in vector DB
4. Retrieve relevant chunks
5. Feed them into an LLM


## 🔹 Visual Diagram
```
Text → Chunking → Embeddings → Vector Store → Retrieval → LLM Output
```


## 🧪 Step 1 — Chunk a Text Document


````python
text = """
Machine learning enables systems to learn from data.
It is used in recommendation systems, robotics, medicine, and finance.
Deep learning uses neural networks with many layers.
"""

chunks = text.split(".")
chunks = [c.strip() for c in chunks if c.strip()]
chunks
````

## 🧪 Step 2 — Generate Embeddings (Simulated)
Here we simulate embeddings with random vectors.


````python
import numpy as np

embeddings = {c: np.random.rand(5) for c in chunks}
embeddings
````

## 🧪 Step 3 — Simple Retrieval Function
We compute cosine similarity and retrieve closest chunks.


````python
def cosine(a, b):
    return np.dot(a,b) / (np.linalg.norm(a)*np.linalg.norm(b))

def retrieve(query_embedding, k=1):
    scored = []
    for chunk, emb in embeddings.items():
        scored.append((chunk, cosine(query_embedding, emb)))
    return sorted(scored, key=lambda x: x[1], reverse=True)[:k]

# simulate query
query_emb = np.random.rand(5)
retrieve(query_emb, k=2)
````

## 🧪 Exercise 6 — Build a Simple Retriever Class
Create a Python class:

```python
class SimpleRetriever:
    def __init__(self, embeddings_dict):
        pass

    def retrieve(self, query, k=2):
        pass
```

Tasks:
- Simulate embeddings
- Implement cosine similarity
- Return top-k matches


````python
# TODO: Build the SimpleRetriever class
import numpy as np

class SimpleRetriever:
    def __init__(self, embeddings_dict):
        # embeddings_dict: {text: embedding}
        self.embeddings = embeddings_dict

    def cosine(self, a, b):
        return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

    def retrieve(self, query, k=2):
        # simulate query embedding
        query_embedding = np.random.rand(5)

        scores = []
        for text, emb in self.embeddings.items():
            score = self.cosine(query_embedding, emb)
            scores.append((text, score))

        # return top-k results
        return sorted(scores, key=lambda x: x[1], reverse=True)[:k]
````

## 🧪 Exercise 8 — LangChain Sequential Chain
Build a chain with the following flow:
Input → Summarize → Extract Keywords → Final Output

**Requirements:**
- Use PromptTemplate
- Use two LLMChains
- Use SimpleSequentialChain


````python
# TODO: Implement the sequential chain
from langchain.prompts import PromptTemplate
from langchain.llms import OpenAI
from langchain.chains import LLMChain, SimpleSequentialChain

# Initialize LLM
llm = OpenAI(temperature=0)

# Prompt 1: Summarize
summary_prompt = PromptTemplate(
    input_variables=["text"],
    template="Summarize the following text:\n{text}"
)

summary_chain = LLMChain(
    llm=llm,
    prompt=summary_prompt
)

# Prompt 2: Extract keywords
keyword_prompt = PromptTemplate(
    input_variables=["text"],
    template="Extract the main keywords from the following text:\n{text}"
)

keyword_chain = LLMChain(
    llm=llm,
    prompt=keyword_prompt
)

# Sequential chain
chain = SimpleSequentialChain(
    chains=[summary_chain, keyword_chain],
    verbose=True
)

# Example input
chain.run("Machine learning is a field of artificial intelligence that allows systems to learn from data.")

````
